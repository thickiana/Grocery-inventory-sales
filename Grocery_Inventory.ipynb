```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/Grocery_Inventory new v1.csv')
df.head(5)
```

```python
df.drop(columns=['p'], inplace=True)
df.head()
```

```python
df.info()
```

```python
df.rename(columns={'cost':'Cost','profit':'Profit','Catagory':'Category','percentage':'profit margin'},inplace=True)
```

```python
df.info()
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
df.dropna(subset=['Category'],inplace=True)
print(df.isnull().sum())
```

```python
print("\nDuplicate rows:")
print(df[df.duplicated()])
```

```python
df.rename(columns={'Catagory':'Category','cost':'Cost','profit':'Profit',},inplace =True)
print(df.columns)
```

```python
print("\nSummary statistics:")
print(df.describe().T)
```

```python
categorical_vars = df.select_dtypes(include=['object', 'category']).columns.tolist()
numerical_vars = df.select_dtypes(include=['number']).columns.tolist()

print("Categorical Variables:", categorical_vars)
print("Numerical Variables:", numerical_vars)
```

```python
# Univariate analysis for numerical variable
numerical_vars
```

```python
# UNIVARIATE ANALYSIS  Detecting outliers using boxplots
plt.figure(figsize=(10, 6))
sns.boxplot(data=df.select_dtypes(include=np.number))
plt.xticks(rotation=90)
plt.title("Outliers Detection using Boxplot")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.kdeplot(df["Sales"], fill=True, color="blue")
plt.title("KDE Plot of Sales")
plt.show()
```

```python
categorical_vars
```

```python
columns_to_drop =['Supplier_Name','Warehouse_Location','Status','Product_ID','Supplier_ID','Date_Received','Last_Order_Date','Expiration_Date']
df.drop(columns=columns_to_drop, inplace=True)

df.head()
```

```python
sns.countplot(x='Category', data=df)  # or sns.countplot(y='Category', data=df) for horizontal bars
plt.title('Count of Category')  # Customize the title as needed
plt.xlabel('Category')  # Customize the x-axis label
plt.ylabel('Count')  # Customize the y-axis label
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
# bivariate analysis scatter plot(NUMERICAL FEATURES)
plt.figure(figsize=(8, 4))
sns.scatterplot(x='Sales', y='Sales_Volume',hue='Category', data=df)
plt.title('Scatter Plot:sales vs Sales Volume')
plt.xlabel('Sales')
plt.ylabel('Sales Volume')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.scatterplot(x='Sales', y='Profit',hue='Category', data=df)
plt.title('Scatter Plot:Salest vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()
```

```python
df['Profit'] = pd.to_numeric(df['Profit'], errors='coerce')

plt.figure(figsize=(10, 6))  # Adjust figure size as needed
sns.boxplot(x='Category', y='Profit', data=df)
plt.title('Profit Distribution by Category')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels if needed
plt.tight_layout()
plt.show()
```

```python
df.groupby("Category").sum()["Cost"].plot(kind="bar");plt.title("Total cost by category")
```

```python
df.groupby("Category").sum()["Profit"].plot(kind="bar");plt.title("Total profit by category")
```

```python
df.Product_Name.value_counts()
```

```python
plt.figure(figsize=(10, 6))
category_counts = df['Category'].value_counts().reset_index()
category_counts.columns = ['Category', 'Inventory_Turnover_Count']


sns.barplot(x='Category', y='Inventory_Turnover_Count', data=category_counts)

plt.title('Number of Inventory Turnovers by Category')
plt.xlabel('Category')
plt.ylabel('Inventory Turnover Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python

# Group data by 'Category' and sum 'Sales_Volume'
category_sales_volume = df.groupby('Category')['Sales_Volume'].sum().reset_index()

# Create the bar plot
plt.figure(figsize=(10, 6))  # Adjust figure size as needed
sns.barplot(x='Category', y='Sales_Volume', data=category_sales_volume)
plt.title('Sales Volume by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales Volume')
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels if needed
plt.tight_layout()
plt.show()
```

```python

category_sales = df.groupby('Category')['Sales'].sum().reset_index()

# Create the bar plot
plt.figure(figsize=(10, 6))  # Adjust figure size as needed
sns.barplot(x='Category', y='Sales', data=category_sales)
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels if needed
plt.tight_layout()
plt.show()
```

```python
#heatmap correlation
plt.figure(figsize=(10, 6))
sns.heatmap(df.select_dtypes(include=np.number).corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(df[["Stock_Quantity", "Reorder_Level", "Reorder_Quantity", "Sales_Volume","Sales","Cost","Inventory_Turnover_Rate"]])
plt.show()
```

```python


# Group data by 'Product_Name' and sum 'Sales_Volume' to get total orders
product_orders = df.groupby('Product_Name')['Sales_Volume'].sum().reset_index()

# Sort by total orders in descending order
product_orders = product_orders.sort_values(by=['Sales_Volume'], ascending=False)

# Create the bar chart using seaborn
plt.figure(figsize=(12, 6))  # Adjust figure size as needed
sns.barplot(x='Product_Name', y='Sales_Volume', data=product_orders)
plt.title('Most Ordered Products')
plt.xlabel('Product Name')
plt.ylabel('Total Orders (Sales Volume)')
plt.xticks(rotation=90, ha='right')  # Rotate x-axis labels for readability
plt.tight_layout()
plt.show()
```
